# The OF-DFT KS equation


In orbital-free DFT

$$T_s[n]=T_{vW}[n] + {T_P[n]}$$

<div style="text-align:center; max-width:900px; margin:0 auto; border:1px solid #e6c200; background:#fcf8e3; padding:14px 18px;">
<b>von Weizsäcker vs Pauli (orbital-free)</b><br><br>
The <b>von Weizsäcker</b> piece is written explicitly with the density; the <b>Pauli</b> piece collects the rest of the non-interacting kinetic energy and is <b>not known exactly</b> as a simple density functional.
</div>

$$
T_{\mathrm{vW}}[n] = -\frac{1}{2}\big\langle \sqrt{n} \,\big|\, \nabla^2 \,\big|\, \sqrt{n} \big\rangle = -\frac{N}{2}\big\langle \phi \,\big|\, \nabla^2 \,\big|\, \phi \big\rangle
$$

$T_{\mathrm{P}}[n]$ is the <b>(unknown) Pauli kinetic energy</b> such that $T_s[n] = T_{\mathrm{vW}}[n] + T_{\mathrm{P}}[n]$.


<div style="text-align:center; max-width:900px; margin:0 auto; border:1px solid #3c763d; background:#dff0d8; padding:14px 18px;">
<b>OF Lagrangian (sketch)</b><br><br>
Write the orbital-free Lagrangian with kinetic, Hartree, XC, external, and chemical-potential terms; then impose <b>stationarity</b> with respect to either the <b>density</b> or the <b>normalized square-root field</b> (Boson variable).
</div>

$$
\mathcal{L}_{\mathrm{OF}}[\phi] = -\frac{N}{2}\big\langle \phi \,\big|\, \nabla^2 \,\big|\, \phi \big\rangle + T_{\mathrm{P}}[n] + E_{\mathrm{H}}[n] + E_{\mathrm{xc}}[n] + \int v_{\mathrm{ext}}(\mathbf{r})\, n(\mathbf{r})\, d\mathbf{r} - \mu \left( \int n(\mathbf{r})\, d\mathbf{r} - N \right)
$$

Then impose stationarity with respect to either $n(\mathbf{r})$ or $\phi(\mathbf{r}) = \dfrac{1}{\sqrt{N}}\sqrt{n(\mathbf{r})}$.

# What is the physical meaning of KS and OF orbitals?

- Generally KS or HF orbitals have no rigorous physical meaning
- The HOMO is different: HF has <a href="https://www.sciencedirect.com/science/article/abs/pii/S0031891434900112?via%3Dihub">Koopmans' theorem</a> and KS has <a href="https://journals.aps.org/prb/abstract/10.1103/PhysRevB.18.7165">Janak's theorem</a> relating it to the IP.
- What about the virtuals? The LUMO? ... no rigorous physical meaning.

# Can the KS orbitals tell us something useful?

<center>
    <div class="alert alert-success">Yes! In a few ways.</div>
</center>

### If you are interested in ionization energies:
- GW $$\varepsilon_i \to \varepsilon_i + \langle \phi_i | \hat \Sigma(\varepsilon_i) - v_{xc} | \phi_i \rangle$$

### If you are interested in excitation energies:

- TDDFT $$\varepsilon_a-\varepsilon_i \to (\varepsilon_a-\varepsilon_i) + \left\langle \phi_i\phi_a \bigg| \frac{1}{|r-r'|}+f_{xc}(r,r',\varepsilon_a-\varepsilon_i) \bigg| \phi_i\phi_a \right\rangle+\cdots$$

where $i,j,k,l,\ldots \in \text{occupied}$ and $a,b,c,d,\ldots \in \text{virtuals}$.

# PART II: TDDFT based on KS-DFT and OF-DFT

# KS-TDDFT and OF-TDDFT equations :

<br>
<br>

### KS-TDDFT

$$
-\frac{1}{2}\nabla^2 \phi_i(\mathbf{r},t) + \textcolor{red}{v_s(\mathbf{r},t)}\,\phi_i(\mathbf{r},t)
= i\,\frac{\partial \phi_i(\mathbf{r},t)}{\partial t}
$$

$$
n(\mathbf{r},t) = \sum_i n_i\, \lvert \phi_i(\mathbf{r},t) \rvert^2,
\qquad N = \sum_i n_i,
\quad \text{with } \{n_i\} \text{ time-independent.}
$$

<br>

### OF-TDDFT

$$
-\frac{1}{2}\nabla^2 \phi(\mathbf{r},t) + \textcolor{blue}{v_B(\mathbf{r},t)}\,\phi(\mathbf{r},t)
= i\,\frac{\partial \phi(\mathbf{r},t)}{\partial t}
$$

$$
n(\mathbf{r},t) = N\, \lvert \phi(\mathbf{r},t) \rvert^2
$$

with, for example,

$$
\phi(\mathbf{r},t) = \frac{1}{\sqrt{N}}\,\sqrt{n(\mathbf{r},t)}\, e^{\,i S(\mathbf{r},t)}.
$$


<center>
    <div class="alert alert-success">The KS-TDDFT equations are the result of the <a href="https://journals.aps.org/prl/abstract/10.1103/PhysRevLett.52.997">Runge-Gross theorem</a>, the follow-up <a href="https://journals.aps.org/prl/abstract/10.1103/PhysRevLett.82.3863">van Leuwen theorem</a> and considerations by <a href="https://journals.aps.org/pra/abstract/10.1103/PhysRevA.77.062511">Vignale</a>.</div>
</center>

<center>
    <div class="alert alert-success">The OF-TDDFT equations rest on <a href="https://journals.aps.org/prb/abstract/10.1103/PhysRevB.103.245102">similar footing</a>.</div>
</center>

# Integrating the TDDFT equations in real time (rt)

The main idea: use a <b>propagator</b>!
$$
\phi_i(t+dt) = \hat U(t,t+dt)\, \phi_i(t)
$$

It is formally defined as follows for time-independent Hamiltonians:
$$
\hat U(t,t+dt) = e^{-i\hat H\, dt}
$$

Where for KS-TDDFT one would like to write

$$
\textcolor{red}{
\hat H = \hat h_{\mathrm{KS}} = -\frac{1}{2}\nabla^2 + v_s(\mathbf{r},t)
}
$$


For OF-TDDFT

$$
\textcolor{blue}{
\hat H = \hat h_{\mathrm{OF}} = -\frac{1}{2}\nabla^2 + v_B(\mathbf{r},t)
}
$$

# There are complications

- the exponential function is of difficult implementation because it implies diagonalization of the Hamiltonian (almost always not a viable option), or a Taylor expansion $$ e^{M}=1+M+\frac{M^2}{2}+\cdots $$

- The Hamiltonian (KS and OF) contains time-dependent potentials, for example the externally applied potential, $v_{appl}(r,t)$.

# Time-dependence of $\hat H$

The adiabatic approximation for KS-TDDFT:
$$
v_{xc}(r,t) = v_{xc}[n(r,t)](r)
$$
uses the ground state XC functional and potential and evaluates them with the time-dependent density, $n(r,t)$.

- The time-dependent xc potential is <b>not</b> defined in this way

- The adiabatic approximation is perhaps the best we can do. <b>But</b> together with the self-interaction error,  it leads to some shortcomings:
    - <a href="https://iopscience.iop.org/article/10.1088/1361-648X/aa836e/meta">charge-transfer excitations are generally too low in energy</a>
    - predicted optical gaps are usually too low (e.g., <a href="https://journals.aps.org/prb/abstract/10.1103/PhysRevB.73.094204">PBE water: 6.5 eV vs 8.5 eV</a>, see also <a href="https://pubs.acs.org/doi/10.1021/acs.jpclett.7b02212">this paper</a>.)
    - <a href="https://doi.org/10.1016/j.chemphys.2011.03.020">neglect of double excitations (e.g., 1,3-butene)</a>

- There are fixes <b>but you must evaluate them carfully before use</b>.
    - use of hybrids
    - use of GGA+U
    - use of long-range corrected xc potentials <a href="https://pubs.aip.org/aip/jcp/article-abstract/112/3/1344/184896/Molecular-calculations-of-excitation-energies-and">(LB94, SAOP)</a>
    - other functionals (Koopman's, etc)

# Real-time TDDFT

<center>
    <div class="alert alert-success">Real-time TDDFT refers to the integration of the KS/OF-TDDFT equations discretizing time and propagating the occupied orbitals with specialized propagators.</div>
</center>

In rt-TDDFT, propagators are iteratively applied, e.g.:

``` python
phi[0]={ground state occupied orbital(s)}
for i in range(1,N_max):
    phi[i] = U(t,t+dt) phi[i-1]
```

$\hat U(t,t+dt)$ needs to be approximated

Simplest is Taylor expansion of $\hat U=e^{-i\hat H(t) dt}$: $U(t,t+dt)\simeq 1 - i \hat H dt$

## Better idea ... 

<center><img src="../figures/science/cn_time_line.png" width=1000 /></center>

# Crank-Nicolson (CN) propagator

The most common propagator to use is the Crank-Nicolson (CN) propagator:
$$
\hat U(t,t+dt) \simeq \frac{1-i\hat H(t+\frac{dt}{2})\frac{dt}{2}}{1+i\hat H(t+\frac{dt}{2})\frac{dt}{2}}
$$
which propagates with an error of the order of $dt^3$ and is stable regardless of the initial conditions.

- CN conserves the norm, $\langle \phi_i(t) | \phi_i(t)\rangle = \langle \phi_i(t+dt) | \phi_i(t+dt)\rangle = 1$.

- CN may require <b>predictor-corrector</b> steps to better evaluate $\hat H(t+\frac{1}{2}dt)$ and deliver a physical electron dynamics.

- There are many other propagators (ETRS, ...). <a href="https://pubs.aip.org/aip/jcp/article-abstract/121/8/3425/464718/Propagators-for-the-time-dependent-Kohn-Sham?redirectedFrom=fulltext">Check out this nice paper</a> and <a href="https://www.octopus-code.org/documentation/main/manual/calculations/time-dependent/">this nice intro</a> in the OCTOPUS manual.

# Challenge 6

- Derive the Crank–Nicolson (CN) propagator: express $\phi\left(t + \frac{\Delta t}{2}\right)$ from $\phi(t)$ (forward propagation) and $\phi(t+\Delta t)$ (backward propagation), using Taylor expansions in $\Delta t$.
- **Optional:** show that the CN propagator conserves the norm. *Hint:* check the norm directly; the propagator is unitary, so the adjoint equals its inverse.

# What is real-time TDDFT useful for?

- Optical spectra (KS):
    - <a href="https://journals.aps.org/prb/abstract/10.1103/PhysRevB.73.094204">Liquid water</a> and <a href="https://pubs.acs.org/doi/10.1021/acs.jpclett.7b02212">another liquid water</a>
    - <a href="https://pubs.acs.org/doi/10.1021/acsnanoscienceau.2c00061">Metallic systems (plasmons)</a>    

- Optical spectra (OF):
    - <a href="https://journals.aps.org/prb/pdf/10.1103/PhysRevB.104.235110">Nanoparticles</a> and <a href="https://pubs.acs.org/doi/full/10.1021/acs.nanolett.2c01399">other plasmonic systems</a>

- Electron-nuclear dynamics in non-equilibrium systems (KS):
    - <a href="https://www.nature.com/articles/s41524-023-01157-7">Materials stopping power (alpha particles, protons, etc)</a>
    - <a href="https://www.pnas.org/doi/full/10.1073/pnas.1518224112">Polaritonic systems</a>
    - <a href="https://pubs.acs.org/doi/10.1021/acs.jpca.3c03709">Non-equilibrium electron nuclear dynamics (Ehrenfest) of charge-transfer dynamics</a>

- Electron-nuclear dynamics in non-equilibrium systems (OF):
    - <a href="https://journals.aps.org/prl/abstract/10.1103/PhysRevLett.121.145001">Warm dense matter</a>

# Linear-response TDDFT

**Idea:** instead of propagating the system in **real time**, study **small harmonic perturbations** of the external potential and the **linear** change of the density at frequency $\omega$.

This gives **excitation energies** and **spectra** (e.g. Casida formalism for finite systems / molecular electronics; analogous constructions in extended systems).

Longer reference: [DFTpy docs — *Introduction to DFT* (Linear response)](https://dftpy.rutgers.edu/ofdft.html#linear-response-theory).


## Induced density and response function

For a weak perturbing potential $\delta v_{\mathrm{appl}}(\mathbf{r},\omega)$, the **linear** change of the density satisfies

$$
\delta n(\mathbf{r},\omega) = \int \chi(\mathbf{r},\mathbf{r}',\omega)\, \delta v_{\mathrm{appl}}(\mathbf{r}',\omega)\, d\mathbf{r}'
$$

where $\chi$ is the **interacting** response kernel (frequency-dependent polarizability / density–density response).


## KS vs OF reference: same interacting $\delta n$, two routes

Working to **linear order**, the same $\delta n$ can be expressed through the **Kohn–Sham** or **bosonic (OF)** non-interacting responses:

$$
\delta n(\mathbf{r},\omega) = \int \chi_s(\mathbf{r},\mathbf{r}',\omega)\, \delta v_s(\mathbf{r}',\omega)\, d\mathbf{r}'
= \int \chi_B(\mathbf{r},\mathbf{r}',\omega)\, \delta v_B(\mathbf{r}',\omega)\, d\mathbf{r}'
$$

The **KS** potential variation splits as

$$
\delta v_s(\mathbf{r},\omega) = \delta v_H[n](\mathbf{r},\omega) + \delta v_{\mathrm{xc}}[n](\mathbf{r},\omega) + \delta v_{\mathrm{appl}}(\mathbf{r},\omega)
$$

with Hartree piece $\propto \int \dfrac{\delta n(\mathbf{r}',\omega)}{|\mathbf{r}-\mathbf{r}'|}\, d\mathbf{r}'$ and an **xc kernel** $f_{\mathrm{xc}}$.

The **OF / Boson** potential variation adds the **Pauli** channel:

$$
\delta v_B(\mathbf{r},\omega) = \delta v_s(\mathbf{r},\omega) + \delta v_P[n](\mathbf{r},\omega)
$$


## Dyson / Casida viewpoint (compact)

Relating the **interacting** $\chi$ to the **KS** $\chi_s$ leads to a **Dyson**-type screening equation (schematically):

$$
-\chi^{-1} = -\chi_s^{-1} + \frac{1}{|\mathbf{r}-\mathbf{r}'|} + f_{\mathrm{xc}}(\mathbf{r},\mathbf{r}',\omega)
$$

The **OF** route replaces $\chi_s$ by $\chi_B$ and adds a **Pauli kernel** $f_P$:

$$
-\chi^{-1} = -\chi_B^{-1} + \frac{1}{|\mathbf{r}-\mathbf{r}'|} + f_{\mathrm{xc}}(\mathbf{r},\mathbf{r}',\omega) + f_P(\mathbf{r},\mathbf{r}',\omega)
$$

The Dyson equation turns into a matrix equation, known as the **Casida equation**, given by:

$$
\begin{pmatrix} A_{iajb}(\omega) & B_{iajb}(\omega) \\ B_{iajb}(\omega) & A_{iajb}(\omega) \end{pmatrix} \begin{pmatrix} X_{ia} \\ Y_{ia} \end{pmatrix} = \omega \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} X_{ia} \\ Y_{ia} \end{pmatrix}
$$

with $\omega_{ia} = \varepsilon_a - \varepsilon_i$,

$$
A_{iajb}(\omega) = \delta_{ij}\delta_{ab}\,\omega_{ia} + \big\langle \phi_i \phi_a \big| K(\mathbf{r},\mathbf{r}',\omega) \big| \phi_j \phi_b \big\rangle, \qquad
B_{iajb}(\omega) = \big\langle \phi_i \phi_a \big| K(\mathbf{r},\mathbf{r}',\omega) \big| \phi_j \phi_b \big\rangle
$$

In **KS** TDDFT, $K = 1/|\mathbf{r}-\mathbf{r}'| + f_{\mathrm{xc}}$; in **OF** TDDFT, $K$ also includes $f_P$. Excitation energies $\Omega_n$ and transition amplitudes follow from the eigenvectors $(X_{ia}, Y_{ia})$.

## OF vs KS

**Plasmons:** collective modes appear as combinations of many particle–hole pairs in the **KS** picture, but can be more **direct** in a **bosonic** formulation; practical kernels (e.g. TF–vW in LR) shift peaks vs KS-TDDFT, and **nonadiabatic** effects (e.g. **JP** Pauli term in DFTpy) matter for **Landau damping** at the documented level of theory.


# Thank you!

# Ready for some hands-on work?

- KS-TDDFT with QEpy:
    - rt-tddft: [`tddft_c6h6.ipynb`](../tutorials/qepy/rt-tddft/tddft_c6h6.ipynb)
    - lr-tddft:  [`tddfpt_qepy.ipynb`](../tutorials/qepy/lr-tddft/tddfpt_qepy.ipynb)

- OF-TDDFT with DFTpy:
    - rt-tddft: [`td-ofdft-tutorial.ipynb`](../tutorials/dftpy/td-ofdft-tutorial.ipynb) · [DFTpy online](https://dftpy.rutgers.edu/tutorials/jupyter/td-ofdft-tutorial.html) · [Colab](https://colab.research.google.com/github/Quantum-MultiScale/DFTpy/blob/dev/examples/notebooks/of-ofdft-tutorial.ipynb)
    - lr-tddft: [`lr-ofdft-tutorial.ipynb`](../tutorials/dftpy/lr-ofdft-tutorial.ipynb) · [DFTpy online](https://dftpy.rutgers.edu/tutorials/jupyter/lr-ofdft-tutorial.html) · [Colab](https://colab.research.google.com/github/Quantum-MultiScale/DFTpy/blob/dev/examples/notebooks/lr-ofdft-tutorial.ipynb)
